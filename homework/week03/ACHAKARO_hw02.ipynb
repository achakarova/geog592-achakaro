```python
import csv
tabela = "tabela1613.csv"
with open(tabela, "r", errors="ignore") as file: #bc errors are from special symbols
    reader = csv.reader(file) #reader reads everything as a list (no problems), while dictreader makes the column names the header and causes problems
    rows = list(reader)

print("Total rows:", len(rows))
for row in rows[:6]:
    print(row)
```

```python
rows_keep = rows[5:5549]
print("After deleting:", len(rows_keep))
print(rows_keep[0])

rows_pa = []
for row in rows_keep:
    if "(PA)" in row[2]:
        rows_pa.append(row)
print("PA rows only:", len(rows_pa))
print(rows_pa[0])
```

```python
selected_columns = []
for row in rows_pa:
    new_row = {"Cod": row[1], "Acai": row[5]} #Make new dict new_row
    selected_columns.append(new_row)
print(selected_columns[0])
```
