11. Would EPSG:4326 be a good CRS for measuring distance in feet

No, EPSG:4326 uses geographic coordinates in degrees rather than feet, so the data should be projected into a proper CRS first

```python
#12. Create stops_nc by reprojecting stops to EPSG:2264
stops_nc = stops.to_crs(epsg=2264)
stops_nc.head()
```

```python
#13. Create an area_sqft column in parks_nc using polygon area
parks_nc = parks.to_crs(epsg=2264)
parks_nc["area_sqft"] = parks_nc.geometry.area
parks_nc[["NAME", "area_sqft"]]
```

```python
#14. Calculate distance from every stop to one polygon named park_1
park_1 = parks_nc.geometry.iloc[0]
stops_nc["distance_ft"] = stops_nc.distance(park_1)
stops_nc[["NAME", "distance_ft"]]
```

15. If park_1 is a polygon, does .distance(park_1) measure distance to the polygon centroid

No, it measures the shortest distance to the polygon itself, which is zero for points inside or on the polygon

16. Explain each spatial predicate

within: the geometry lies completely inside another geometry  
contains: the geometry contains the other geometry  
intersects: the geometries share at least one point  
touches: their boundaries meet without their interior overlapping

17. In gpd.sjoin(parks, urban, predicate="within"), which geometry is tested as within which

18. A park crosses an urban-area boundary, would within or intersects be True

within would be False if part of the park lies outside the urban area, while intersects would be True because the two polygons overlap

```python
#19. Count how many stops are inside each park polygon
matches = gpd.sjoin(stops, parks[["NAME", "geometry"]], how="inner", predicate="within")
stop_counts = matches.groupby("index_right").size()
parks["stop_count"] = stop_counts.reindex(parks.index, fill_value=0).astype(int)
parks[["NAME", "stop_count"]]
```

20. What is the purpose of gpd.sjoin()

It joins attributes from two GeoDataFrames based on a spatial relationship between their geometries

21. Why is a spatial join different from a regular pandas join on a common ID field

A spatial join matches features by their geometric relationship rather than matching values in a shared ID column

```python
#22. Keep bus stops that fall inside park polygons using a spatial join
stops_in_parks = gpd.sjoin(stops, parks[["NAME", "geometry"]], how="inner", predicate="within")
stops_in_parks
```

23. How could you count how many bus stops were matched to each park

24. Why can polygon-to-polygon intersects produce multiple output rows

One polygon can match several polygons in the other layer, producing one row for each matching pair, but if each layer truly contains exactly one polygon, there can be at most one matched row

```python
#25a. Find parks within 500 feet of a bus stop using distance and apply
near_stop = parks_nc.geometry.apply(lambda park: stops_nc.distance(park).min() <= 500)
parks_within_500ft = parks_nc[near_stop].copy()
parks_within_500ft[["NAME", "geometry"]]
```

```python
#25b. Find parks within 500 feet of a bus stop using buffers
stop_buffers = stops_nc.buffer(500)
buffer_area = stop_buffers.union_all()
parks_near_buffers = parks_nc[parks_nc.intersects(buffer_area)].copy()
parks_near_buffers[["NAME", "geometry"]]
```
