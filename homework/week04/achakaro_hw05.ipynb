28. Explain the difference between equal interval and quantile classification

Answer: Equal interval classification uses classes with equal numeric width, while quantile classification puts roughly the same number of observations in each class

```python
values = [5, 12, 3, 20] #29. Short Python loop that prints only values greater than 10

for value in values:
    if value > 10:
        print(value)

s = pd.Series(values) #Create a Series called s using the same values
s > 10 #Pandas Boolean condition
```
