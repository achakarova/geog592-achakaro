```python
import csv
filename = "2026-09-11T01_25_54.827Z-daily.csv" #USGS Daily Streamflow https://api.waterdata.usgs.gov/download
#monitoring stations: USGS-02097517,USGS-02068500,USGS-03453500
#parameter: 00060
#date range: 01/01/2021 - 12/31/2023

with open(filename, "r") as file:
    reader = csv.DictReader(file)
    rows = list(reader)

print(f"Total rows: {len(rows)}")
print(rows[0])
print(rows[0].keys())
```

```python
selected_columns = []

for row in rows:
    new_row = {
        "Station": row["monitoring_location_id"],
        "Date": row["time"],
        "Discharge": row["value"],
        "Longitude": row["x"],
        "Latitude": row["y"]
    }
    selected_columns.append(new_row)

print(selected_columns[0])
```

```python
missing_values = []

for row in selected_columns:
    if row["Station"] == "" or row["Date"] == "" or row["Discharge"] == "" or row["Longitude"] == "" or row["Latitude"] == "":
        missing_values.append(row)

print(f"Rows with missing values: {len(missing_values)}")
```

```python
non_numeric = []

for row in selected_columns:
    try:
        float(row["Discharge"])
    except ValueError:
        non_numeric.append(row["Discharge"])

print(f"Non-numeric discharge values: {non_numeric}")
```

```python
for row in selected_columns:
    row["Discharge"] = float(row["Discharge"])
    row["Longitude"] = float(row["Longitude"])
    row["Latitude"] = float(row["Latitude"])

print(selected_columns[0])
print(type(selected_columns[0]["Discharge"]))
print(type(selected_columns[0]["Longitude"]))
print(type(selected_columns[0]["Latitude"]))
```

```python
fieldnames = selected_columns[0].keys()

with open("streamflow_clean.csv", "w", newline="") as file:
    writer = csv.DictWriter(file, fieldnames=fieldnames)
    writer.writeheader()
    writer.writerows(selected_columns)
```
